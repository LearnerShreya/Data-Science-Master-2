
## 📌 **Support Vector Classifier (SVC)**

---

### ✅ **What is Support Vector Classifier (SVC)?**

Support Vector Classifier is a **supervised learning model** derived from the **Support Vector Machine (SVM)** framework, specifically used for **classification tasks**.

SVC aims to find the **optimal hyperplane** that separates data into distinct classes with the **maximum margin**. It works well for both **binary** and **multi-class** classification problems.

---

### ⚙️ **How SVC Works:**

1. **Maximizes Margin**: It searches for a hyperplane that maximizes the margin between the nearest data points of different classes (called support vectors).
2. **Uses Kernels**: When data is not linearly separable, it uses the **kernel trick** to project data into higher dimensions.
3. **Optimizes Objective**: The classifier minimizes classification error while maintaining a wide margin using a regularization parameter `C`.

---

### 🧠 **Mathematical Formulation:**

For **linearly separable data**, the objective is to:

$$
\min_{w, b} \frac{1}{2} \|w\|^2
$$

Subject to:

$$
y_i (w \cdot x_i + b) \geq 1
$$

For **non-linearly separable data**, we allow some misclassification using slack variables $ \xi_i $:

$$
\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} \xi_i
$$

Subject to:

$$
y_i (w \cdot x_i + b) \geq 1 - \xi_i, \quad \xi_i \geq 0
$$

Where:
- $ x_i $: Feature vector  
- $ y_i $: Class label (+1 or -1)  
- $ w $: Weight vector  
- $ b $: Bias  
- $ C $: Regularization parameter controlling the trade-off between margin and misclassification  

---

### 🔍 **SVC vs SVM:**

| Feature | SVM | SVC |
|--------|-----|-----|
| Scope | General term | Specific implementation for classification |
| Uses | Classification, Regression, Outlier Detection | Only classification |
| Library Name | Not used directly | `SVC` class in `scikit-learn` |

---

### 💡 **Important Parameters in SVC (Scikit-learn):**

| Parameter | Description |
|----------|-------------|
| `C` | Regularization parameter: smaller `C` gives a wider margin, allowing more misclassifications. |
| `kernel` | Specifies the kernel type: `'linear'`, `'poly'`, `'rbf'`, `'sigmoid'`, or a custom function. |
| `degree` | Degree of the polynomial kernel (if used). |
| `gamma` | Controls the influence of each training example (used in `rbf`, `poly`, `sigmoid`). |
| `decision_function_shape` | `'ovo'` (one-vs-one) or `'ovr'` (one-vs-rest) for multi-class classification. |

---

### 🔧 **Kernels in SVC:**

#### Common Kernel Functions:

| Kernel | Formula | Use Case |
|--------|---------|----------|
| Linear | $$ K(x, y) = x \cdot y $$ | When data is linearly separable |
| Polynomial | $$ K(x, y) = (x \cdot y + c)^d $$ | Captures polynomial relations |
| RBF (Gaussian) | $$ K(x, y) = e^{-\gamma \|x - y\|^2} $$ | For complex, non-linear data |
| Sigmoid | $$ K(x, y) = \tanh(\alpha x \cdot y + c) $$ | Similar to neural networks |

---

### 🧪 **Example – Using SVC in Python:**

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

# Load dataset
iris = datasets.load_iris()
X, y = iris.data, iris.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Create and train model
svc_model = SVC(kernel='rbf', C=1.0, gamma='scale')
svc_model.fit(X_train, y_train)

# Predict
y_pred = svc_model.predict(X_test)
```

---

### ✅ **Advantages of SVC:**

- Works well in **high-dimensional spaces**.
- Effective when there's a **clear margin of separation**.
- Memory-efficient: uses only **support vectors**.
- Supports both **binary and multi-class** classification.

---

### ❌ **Disadvantages of SVC:**

- **Slow** on large datasets.
- **Sensitive** to parameter selection and kernel choice.
- **Less interpretable** compared to decision trees or logistic regression.
