# 🌳 Decision Tree Classifier 

---

## 📌 **1. Introduction**

- **Decision Tree** is a supervised machine learning algorithm.
- Types:
   - Decision Tree Classifier
   - Decision Tree Regressor
- Used for **classification** (mainly) and **regression** problems.
- It splits the data into subsets based on feature values, forming a **tree-like structure**.
- The goal is to create a model that predicts the value of a target variable by learning **simple decision rules** from data features.

---

## 🔍 **2. Real-Life Analogies & Applications**

- Like playing “20 Questions” – keep asking **yes/no** questions until you narrow down to a decision.
- **Applications:**
  - Medical diagnosis (Is it flu or cold?)
  - Loan eligibility in banking
  - Spam email classification
  - Student performance prediction
  - Disease prediction in healthcare
  - Crop recommendation in agriculture

---

## 🌿 **3. Key Terminologies**

| Term                  | Description |
|-----------------------|-------------|
| **Root Node**         | The first split, contains all the data |
| **Decision Node**     | Nodes where data is split |
| **Leaf/Terminal Node**| Final outcome or class |
| **Splitting**         | Dividing a node into two or more |
| **Pruning**           | Removing unnecessary branches to reduce overfitting |
| **Branch**            | A path from decision to result |
| **Depth**             | Number of splits from root to leaf |
| **Impurity**          | Measure of disorder (more impurity = less purity = worse split) |

---

## 🧠 **4. Intuition (How It Works)**

1. Start with the **entire dataset**.
2. At each node, choose the **best feature and threshold** that gives the most effective split.
3. Split the dataset based on this feature.
4. Repeat recursively until:
   - All samples at a node belong to the same class
   - Maximum depth reached
   - No further gain from splitting

> It tries to create **pure groups** step-by-step, where all values belong to the same class.

---

## 🧮 **5. Splitting Criteria (Mathematics)**

### 📘 Entropy (Information Theory)
- Measures **disorder** or uncertainty
- Formula:  
  $$
  Entropy = -\sum_{i=1}^{n} p_i \log_2(p_i)
  $$

### 📗 Information Gain
- Reduction in entropy from parent to child nodes
- Formula:  
  $$
  IG = Entropy(Parent) - \sum_k \frac{|D_k|}{|D|} \cdot Entropy(D_k)
  $$

### 📕 Gini Impurity (used in CART)
- Measures probability of misclassification
- Formula:  
  $$
  Gini = 1 - \sum_{i=1}^{n} p_i^2
  $$

#### ✳️ Gini vs Entropy

| Feature         | Gini Impurity     | Entropy         |
|------------------|------------------|------------------|
| Computation       | Faster           | Slower           |
| Range             | 0 (pure) to 0.5  | 0 (pure) to 1    |
| Bias              | Towards larger classes | Neutral        |

---

## ⚙️ **6. Stopping Criteria**

The tree stops growing when:

- Maximum depth is reached (`max_depth`)
- Minimum number of samples to split is reached (`min_samples_split`)
- Minimum samples in a leaf (`min_samples_leaf`)
- No improvement in information gain
- All data in a node is of the same class

---

## ✂️ **7. Pruning (Controlling Overfitting)**

- Decision Trees can **overfit**, especially deep trees.
  
### a) **Pre-Pruning**
- Stop growing before overfitting
- Use constraints like:
  - `max_depth`
  - `min_samples_split`
  - `max_leaf_nodes`

### b) **Post-Pruning**
- Grow a full tree, then remove branches with **low importance**
- Also called **Cost Complexity Pruning**

## 🧪 **10. Python Implementation (Scikit-learn)**

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load dataset
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target)

# Create and train model
clf = DecisionTreeClassifier(criterion='gini', max_depth=3)
clf.fit(X_train, y_train)

# Predict
y_pred = clf.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
```

## 🖼️ **11. Tree Visualization**

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
plot_tree(clf, filled=True, feature_names=data.feature_names, class_names=data.target_names)
plt.show()
```

## 🧮 5. **Splitting Criteria**

When a Decision Tree grows, it needs to **decide at each step**:  
> _Which feature should be used to split the data to achieve the purest separation?_

To make this decision, we use **impurity measures** like **Entropy** and **Gini Impurity**. Then we evaluate **Information Gain** to see which split gives the best improvement.

---

### 📘 **A. Entropy (Information Theory)**

#### 🔹 What is it?
- Entropy measures **disorder** or **uncertainty** in a dataset.
- A pure dataset (all samples belong to the same class) has **low entropy**.
- A mixed dataset (equal class distribution) has **high entropy**.

#### 🔹 Formula:
$$
Entropy(S) = -\sum_{i=1}^{n} p_i \log_2(p_i)
$$

Where:
- $p_i$ = probability of class `i` in dataset `S`
- `n` = number of classes

#### 🔹 Entropy Range:
- **0** → perfectly pure (e.g., all class A)
- **1** → complete impurity (e.g., 50% A, 50% B in binary case)

---

#### ✅ **Example:**

Suppose we have 10 samples:
- 6 positive (class = 1)
- 4 negative (class = 0)

$$
p_1 = \frac{6}{10} = 0.6,\quad p_0 = \frac{4}{10} = 0.4
$$  
$$
Entropy = -(0.6 \cdot \log_2 0.6 + 0.4 \cdot \log_2 0.4)
        \approx - (0.6 \cdot -0.737 + 0.4 \cdot -1.321)
        \approx 0.971
$$

---

### 📗 **B. Information Gain (IG)**

#### 🔹 What is it?
- **Information Gain** is the **reduction in entropy** after a dataset is split on an attribute.
- Helps determine **which feature is best** to split the dataset.

#### 🔹 Formula:
$$
IG = Entropy(parent) - \sum_{k=1}^{m} \frac{|D_k|}{|D|} \cdot Entropy(D_k)
$$

Where:
- $D$ is the current dataset
- $D_k$ are the child subsets after splitting
- $\frac{|D_k|}{|D|}$ is the weight of each subset
- $m$ = number of children

---

#### ✅ **Example (continued)**

Say we split the 10 samples using a feature, creating:
- **Subset 1:** 4 positive, 0 negative → Entropy = 0
- **Subset 2:** 2 positive, 4 negative →  
  $$
  p_1 = \frac{2}{6},\ p_0 = \frac{4}{6}
  \Rightarrow Entropy \approx - (0.33 \cdot -1.585 + 0.67 \cdot -0.58) \approx 0.918
  $$

Now calculate IG:
$$
IG = 0.971 - \left[ \frac{4}{10} \cdot 0 + \frac{6}{10} \cdot 0.918 \right] = 0.971 - 0.550 = 0.421
$$

The **higher the IG**, the better the split.

---

### 📕 **C. Gini Impurity**

#### 🔹 What is it?
- Measures how often a randomly chosen element would be **incorrectly classified**.
- Used by default in **CART (Classification and Regression Tree)** algorithm.

#### 🔹 Formula:
$$
Gini(S) = 1 - \sum_{i=1}^{n} p_i^2
$$

Where:
- $p_i$ = probability of class `i` in dataset `S`

#### 🔹 Gini Range:
- 0 → pure node (only one class)
- Maximum = $1 - \frac{1}{n}$ → complete impurity

---

#### ✅ **Example with 60% Positives and 40% Negatives:**

$$
Gini = 1 - (0.6)^2 - (0.4)^2 = 1 - 0.36 - 0.16 = 0.48
$$

---

## ✳️ **Gini vs Entropy – Detailed Comparison**

| Aspect                | **Gini Impurity**                          | **Entropy**                              |
|------------------------|--------------------------------------------|-------------------------------------------|
| **Formula**           | $1 - \sum p_i^2$                         | $- \sum p_i \log_2 p_i$                 |
| **Computation**       | ✅ **Faster** (no log)                     | ❌ Slower (uses log)                      |
| **Bias**              | Slightly biased toward **larger classes**  | **Neutral** split decisions               |
| **Range (Binary)**    | [0, 0.5]                                   | [0, 1]                                   |
| **Used In**           | CART (default in scikit-learn)             | ID3, C4.5 algorithms                      |
| **Best For**          | When speed is needed, and class imbalance exists | When interpretability is needed        |

---

## 💡 Which One to Use?

- Both work **similarly in practice**.
- **Gini is preferred** in most modern applications due to **faster computation**.
- **Entropy** is more **theoretically elegant**, but slower.

| 📌 **Purpose**               | Measure **uncertainty/disorder**            | Measure **probability of incorrect label**   | Measure **reduction in impurity** after a split       |

| 📊 **Output Range**         | $$0, 1$$ (Binary classification)          | $$0, 0.5$$ (Binary classification)         | $\geq 0$ (positive real numbers)                    |
| ✅ **Best Value**            | 0 → perfect purity                          | 0 → perfect purity                           | **Higher** = Better split                             |
| 🧮 **Mathematical Complexity**| More complex (uses log)                    | Simpler (no log)                             | Depends on underlying impurity measure (usually Entropy) |
| ⚙️ **Used in Algorithms**    | ID3, C4.5                                   | CART (Classification and Regression Trees)   | ID3, C4.5 (used with Entropy)                         |
| ⚡ **Speed (Computation)**   | Slower due to logarithmic calculation       | Faster – uses only squaring                  | Medium – depends on base measure                      |
| 🎯 **Bias Behavior**         | More neutral / favors balanced splits       | Slightly biased toward **larger** classes    | Follows Entropy bias                                  |
| 🔄 **Interpretability**      | Derived from Information Theory (Shannon)   | More intuitive with misclassification idea   | Measures **actual gain** in class purity              |
| 🧪 **Sensitivity to Classes**| Sensitive to class probabilities            | Less sensitive to changes in class probs     | Direct measure of split quality                       |

## 🧠 **Visual Intuition**

- **Entropy**: Think of flipping a **fair coin** (max disorder = 1) vs a biased coin (less disorder = lower entropy).
- **Gini**: Think of the **chance of mislabeling** if you randomly guess the class.
- **Information Gain**: How much "uncertainty" you **remove** by making a split.

## 🎯 When to Use What?

| **Scenario**                                  | **Best Metric**              | **Why**                                     |
|----------------------------------------------|------------------------------|---------------------------------------------|
| You want fastest training time                | 🟩 **Gini Impurity**         | No logarithm = faster splits                |
| You care about information theory or fairness | 🟪 **Entropy**               | Theoretically grounded and neutral          |
| You want the clearest metric for split quality| 🟦 **Information Gain**      | Directly shows how much impurity reduced    |
| You want to follow standard libraries         | Depends on algorithm         | ID3 → Entropy, CART → Gini                  |
