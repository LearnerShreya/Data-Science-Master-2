### 📚 **Decision Tree for Categorical & Numerical Splits**

---

#### **What is a Decision Tree?**

A **Decision Tree** is a supervised learning algorithm used for classification and regression tasks. It recursively splits the data based on feature values to make decisions, with the goal of minimizing impurity (the disorder within the subsets).

---

### **Splitting Criteria**

The key idea in splitting is to evaluate how "pure" each subset is after the split. **Impurity measures** like **Gini Impurity**, **Entropy**, and **Variance** are used to assess the purity of the resulting subsets.

- **Gini Impurity** (used in classification)
- **Entropy** (used in classification)
- **Variance** (used in regression)

These measures help decide where and how to split the data at each step of the decision tree.

---

### **How Does the Decision Tree Split?**

---

#### **1. Categorical Splits**

Categorical features represent distinct, fixed categories (e.g., **Weather** with values **Sunny**, **Rainy**, **Cloudy**). For categorical features, the tree splits the data based on each **distinct value** of the feature.

##### **Example:**

Target: `Play Tennis` (Yes/No)  
Feature: `Weather` (Sunny, Overcast, Rainy)

**Dataset:**

| ID  | Weather  | Play Tennis |
|-----|----------|-------------|
| 1   | Sunny    | No          |
| 2   | Sunny    | No          |
| 3   | Overcast | Yes         |
| 4   | Rainy    | Yes         |
| 5   | Rainy    | Yes         |
| 6   | Rainy    | No          |
| 7   | Overcast | Yes         |
| 8   | Sunny    | No          |

For the **Weather** feature:
- **Sunny → No, No, No**
- **Overcast → Yes, Yes**
- **Rainy → Yes, Yes, No**

Now, the **Gini Impurity** or **Entropy** for each subset is calculated, and the feature with the best impurity reduction (highest **Information Gain**) is selected.

---

#### **2. Numerical Splits**

Numerical features are continuous and require a **threshold** to create two groups (e.g., **Temperature**, **Age**). The decision tree evaluates different threshold values to split the data.

##### **Example:**

Target: `Play Tennis` (Yes/No)  
Feature: `Temperature`

**Dataset:**

| ID | Temperature | Play Tennis |
|----|-------------|-------------|
| 1  | 85          | No          |
| 2  | 80          | No          |
| 3  | 83          | Yes         |
| 4  | 70          | Yes         |
| 5  | 68          | Yes         |
| 6  | 65          | No          |
| 7  | 64          | Yes         |
| 8  | 72          | No          |

For **Temperature**, we try possible splits based on thresholds between consecutive values:
- Thresholds: `(64+65)/2 = 64.5`, `(65+68)/2 = 66.5`, `(68+70)/2 = 69`, etc.

After evaluating impurity at each potential split, the tree chooses the threshold that best reduces impurity.

---

### **Step-by-Step Example:**

#### **Dataset:**

| ID | Weather  | Temperature | Play Tennis |
|----|----------|-------------|-------------|
| 1  | Sunny    | 85          | No          |
| 2  | Sunny    | 80          | No          |
| 3  | Overcast | 83          | Yes         |
| 4  | Rainy    | 70          | Yes         |
| 5  | Rainy    | 68          | Yes         |
| 6  | Rainy    | 65          | No          |
| 7  | Overcast | 64          | Yes         |
| 8  | Sunny    | 72          | No          |

---

#### **Step 1: Calculate Impurity for Each Feature**

**For `Weather`:**
- **Sunny → No, No, No** → Gini Impurity (or Entropy)
- **Overcast → Yes, Yes** → Gini Impurity (or Entropy)
- **Rainy → Yes, Yes, No** → Gini Impurity (or Entropy)

**For `Temperature`:**
- Calculate the **Gini Impurity** for various thresholds (e.g., Temp ≤ 72).

---

#### **Step 2: Choose the Feature with the Best Split**

After calculating impurity for each feature:
- **Temperature ≤ 72** might show the best reduction in impurity, so we choose it as the first split.

---

### **Final Decision Tree Structure**

Based on the chosen splits:

```plaintext
               [Temperature ≤ 72]
                 /              \
            Yes                  No
         (<= 72)             (> 72)
           /                     \
      [Weather]               [Weather]
      /  |  \                    /     \
   Sunny Overcast Rainy    Sunny  Overcast
   |      |       |          |        |
  No     Yes      Yes       No       Yes
```

---

### **Key Takeaways:**

- **Categorical Splits**: The tree splits based on the categories of the feature (e.g., Weather = Sunny, Rainy).
  
- **Numerical Splits**: The tree splits based on a threshold value (e.g., Temperature ≤ 72).

The decision tree recursively makes splits at each node to reduce impurity, ultimately leading to a final prediction.

---

### **Conclusion:**

A Decision Tree builds itself by selecting features and thresholds that split the data in a way that makes the target variable more predictable. Both **categorical** and **numerical** splits play a crucial role in this process, where impurity measures guide the selection of the best splits.
